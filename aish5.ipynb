```python

class Solution:
    def divide(self, dividend, divisor):
        INT_MAX = 2147483647
        INT_MIN = -2147483648

        if dividend == INT_MIN and divisor == -1:
            return INT_MAX

        negative = (dividend < 0) != (divisor < 0)

        dividend = abs(dividend)
        divisor = abs(divisor)

        quotient = 0

        while dividend >= divisor:
            temp = divisor
            multiple = 1

            while dividend >= temp + temp:
                temp += temp
                multiple += multiple

            dividend -= temp
            quotient += multiple

        if negative:
            quotient = -quotient

        return quotient
```
